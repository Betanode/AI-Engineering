The problem with traditional approach was, we cannot hard code everything, we know more than we can describe, so coding all possible situation was not possible,lol.
We can’t write down if-then rules to cover all situations,
edge cases etc. (i.e., we can’t generalize to new
situations)

To address this issue, we need to use maths. Give computer so many input-output examples and ask computer to make a mathematical model to predict things. This approach is called machine learning.
image.png

Machine learning was a great approach for structure data,like for the data that you can present in numarical format on excel sheet, but for unstrcutured data the story was fully different. ML models were not good for unstrcutured data like Video, Audio, Image.
To use ML on unstructured data, we had to
manually create a better representation of the data
first
image.png



To address this problem, a different
approach was developed – Deep Learning


- Deep Learning can handle unstructured input
data without upfront manual preprocessing!!


- It can automatically extract smart
representations from raw, unstructured data.
- We can simply feed these smart representations
to traditional models like linear regression or
logistic regression and achieve amazing
performance

Deep learning got better because of 3 things
- Better algorithms
- Huge amount of data
- Better computing power

One crazy thing is , we can fit DL model to any sensor. Because all the sensor record the unstuctered data, so process that data using DL model then use ML technique to predict
